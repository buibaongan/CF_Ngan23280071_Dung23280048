```python
import matplotlib.pyplot as plt

class Plot:
    def __init__(self, strategy):
        self.strategy = strategy

    def plot






    def plot_short_time(strategy_name, ticker, time, plot_signals=False):
        #time: how many days to plot 
        if strategy_name == 'mean_reversion':
            plt.figure(figsize=(14, 7))
            plt.plot(strategy.df.index[-time:], strategy.df['Adj Close'][ticker][-time:], label='Price', color='black', alpha=0.6)
            plt.plot(strategy.df.index[-time:], strategy.df['BB_Mean'][ticker][-time:], label='Mean', color='orange', linestyle='--')
            plt.plot(strategy.df.index[-time:], strategy.df['BB_Upper'][ticker][-time:], label='Upper Band', color='green', alpha=0.3)
            plt.plot(strategy.df.index[-time:], strategy.df['BB_Lower'][ticker][-time:], label='Lower Band', color='red', alpha=0.3)
            plt.fill_between(strategy.df.index[-time:], strategy.df['BB_Upper'][ticker][-time:], strategy.df['BB_Lower'][ticker][-time:], color='gray', alpha=0.1)

            if plot_signals:
                buy_signals = strategy.df.index[(strategy.df['Signal'][ticker] == 1) & (strategy.df.index.isin(strategy.df.index[-time:]))]
                sell_signals = strategy.df.index[(strategy.df['Signal'][ticker] == -1) & (strategy.df.index.isin(strategy.df.index[-time:]))]
                plt.scatter(buy_signals, strategy.df['Adj Close'][ticker][buy_signals], marker='^', color='green', label='Buy Signal', s=100)
                plt.scatter(sell_signals, strategy.df['Adj Close'][ticker][sell_signals], marker='v', color='red', label='Sell Signal', s=100)
            plt.title(f'Bollinger Bands for {ticker}')

        elif strategy_name == 'momentum':
            plt.figure(figsize=(14, 7))
            plt.plot(strategy.df.index[-time:], strategy.df['Adj Close'][ticker][-time:], label='Price', color='black', alpha=0.6)
            plt.plot(strategy.df.index[-time:], strategy.df['SMA_50'][ticker][-time:], label='50-day SMA', color='blue', linestyle='--')
            plt.plot(strategy.df.index[-time:], strategy.df['SMA_200'][ticker][-time:], label='200-day SMA', color='red', linestyle='--')
            if plot_signals:
                buy_signals = strategy.df.index[(strategy.df['Signal'][ticker] == 1) & (strategy.df.index.isin(strategy.df.index[-time:]))]
                sell_signals = strategy.df.index[(strategy.df['Signal'][ticker] == -1) & (strategy.df.index.isin(strategy.df.index[-time:]))]
                plt.scatter(buy_signals, strategy.df['Adj Close'][ticker][buy_signals], marker='^', color='green', label='Buy Signal', s=100)
                plt.scatter(sell_signals, strategy.df['Adj Close'][ticker][sell_signals], marker='v', color='red', label='Sell Signal', s=100)
            plt.title(f'SMA Crossover for {ticker}')
        else:
            print("Invalid strategy name. Choose 'mean_reversion' or 'momentum'.")
            return
        

    def plot_full_time(strategy_name, ticker, plot_signals=False):
        # Plot the entire time series
        if strategy_name == 'mean_reversion':
            plt.figure(figsize=(14, 7))
            plt.plot(strategy.df.index, strategy.df['Adj Close'][ticker], label='Price', color='black', alpha=0.6)
            plt.plot(strategy.df.index, strategy.df['BB_Mean'][ticker], label='Mean', color='orange', linestyle='--')
            plt.plot(strategy.df.index, strategy.df['BB_Upper'][ticker], label='Upper Band', color='green', alpha=0.3)
            plt.plot(strategy.df.index, strategy.df['BB_Lower'][ticker], label='Lower Band', color='red', alpha=0.3)
            plt.fill_between(strategy.df.index, strategy.df['BB_Upper'][ticker], strategy.df['BB_Lower'][ticker], color='gray', alpha=0.1)

            if plot_signals:
                buy_signals = strategy.df.index[strategy.df['Signal'][ticker] == 1]
                sell_signals = strategy.df.index[strategy.df['Signal'][ticker] == -1]
                plt.scatter(buy_signals, strategy.df['Adj Close'][ticker][buy_signals], marker='^', color='green', label='Buy Signal', s=100)
                plt.scatter(sell_signals, strategy.df['Adj Close'][ticker][sell_signals], marker='v', color='red', label='Sell Signal', s=100)
            plt.title(f'Bollinger Bands for {ticker}')

        elif strategy_name == 'momentum':
            plt.figure(figsize=(14, 7))
            plt.plot(strategy.df.index, strategy.df['Adj Close'][ticker], label='Price', color='black', alpha=0.6)
            plt.plot(strategy.df.index, strategy.df['SMA_50'][ticker], label='50-day SMA', color='blue', linestyle='--')
            plt.plot(strategy.df.index, strategy.df['SMA_200'][ticker], label='200-day SMA', color='red', linestyle='--')
            if plot_signals:
                buy_signals = strategy.df.index[strategy.df['Signal'][ticker] == 1]
                sell_signals = strategy.df.index[strategy.df['Signal'][ticker] == -1]
                plt.scatter(buy_signals, strategy.df['Adj Close'][ticker][buy_signals], marker='^', color='green')
```
